```python
print("="*60)
print("STEP 2: Encoding Categorical Variables")
print("="*60)

# ========================================
# 2A: Wearable Dataset
# ========================================
print("\n📊 Wearable Dataset:")
print("   Categorical columns: Gender, Diet_Type, Daily_Stress_Level")

# Encode Gender (Label Encoding)
from sklearn.preprocessing import LabelEncoder

df1_clean = df1.copy()

le_gender = LabelEncoder()
df1_clean['Gender_Encoded'] = le_gender.fit_transform(df1_clean['Gender'])

# Encode Diet_Type
le_diet = LabelEncoder()
df1_clean['Diet_Type_Encoded'] = le_diet.fit_transform(df1_clean['Diet_Type'])

# Encode Daily_Stress_Level
stress_mapping = {'Low': 0, 'Medium': 1, 'High': 2}
df1_clean['Daily_Stress_Level_Encoded'] = df1_clean['Daily_Stress_Level'].map(stress_mapping)

print(f"   Gender Mapping: {dict(zip(le_gender.classes_, le_gender.transform(le_gender.classes_)))}")
print(f"   Diet_Type Mapping: {dict(zip(le_diet.classes_, le_diet.transform(le_diet.classes_)))}")
print(f"   Stress Mapping: {stress_mapping}")

# ========================================
# 2B: Student Dataset
# ========================================
print("\n📊 Student Dataset:")
print("   Categorical columns: Student_Type")

# Handle missing values in Student_Type first (fill with mode)
mode_student_type = df2_clean['Student_Type'].mode()[0]
df2_clean['Student_Type'] = df2_clean['Student_Type'].fillna(mode_student_type)

# Encode Student_Type
le_student = LabelEncoder()
df2_clean['Student_Type_Encoded'] = le_student.fit_transform(df2_clean['Student_Type'])

print(f"   Student_Type Mapping: {dict(zip(le_student.classes_, le_student.transform(le_student.classes_)))}")

# ========================================
# 2C: Lifestyle Dataset (Already clean)
# ========================================
print("\n📊 Lifestyle Dataset:")
print("   All columns are already numeric ✅")
df3_clean = df3.copy()
```
